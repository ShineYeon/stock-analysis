```python
import calendar
from datetime import datetime
import psycopg2
from psycopg2.extras import execute_values
import requests
from tqdm.notebook import tqdm
import os
from dotenv import load_dotenv

# ==========================================
# 1. 환경 설정
# ==========================================

load_dotenv()

ECOS_KEY = os.getenv("ECOS_KEY")  # 한국은행 오픈API 인증키 입력

DB_CONFIG = {
    "dbname": os.getenv("DB_NAME"),
    "user": os.getenv("DB_USER"),
    "password": os.getenv("DB_PASSWORD"),  # DB 비밀번호 입력
    "host": os.getenv("DB_HOST"),
    "port": os.getenv("DB_PORT"),
}

STAT_CODE = "731Y001"  # 주요국 통화의 대원화환율
CYCLE = "D"  # 일별 데이터
LANG = "kr"
REQ_TYPE = "json"
SERVICE_NAME = "StatisticSearch"  # 실제 수치 조회 서비스

START_DATE_LIMIT = datetime(1964, 5, 4)
NOW = datetime.now()
```

```python
# ==========================================
# 2. 월별 날짜 구간 생성 함수
# ==========================================
def get_monthly_ranges(start_dt, end_dt):
    ranges = []
    current_year = start_dt.year
    current_month = start_dt.month

    while (current_year, current_month) <= (end_dt.year, end_dt.month):
        _, last_day = calendar.monthrange(current_year, current_month)

        # 시작월(1964-05)인 경우 4일부터 시작
        s_day = start_dt.day if (current_year == start_dt.year and current_month == start_dt.month) else 1
        # 종료월인 경우 오늘 일자까지 제한
        e_day = min(last_day, end_dt.day) if (current_year == end_dt.year and current_month == end_dt.month) else last_day

        start_str = f"{current_year}{current_month:02d}{s_day:02d}"
        end_str = f"{current_year}{current_month:02d}{e_day:02d}"

        ranges.append((start_str, end_str))

        # 다음 달 이동
        if current_month == 12:
            current_year += 1
            current_month = 1
        else:
            current_month += 1

    return ranges
```

```python
# ==========================================
# 3. 메인 수집 및 DB 저장 로직
# ==========================================
date_ranges = get_monthly_ranges(START_DATE_LIMIT, NOW)

# PostgreSQL 연결
conn = psycopg2.connect(**DB_CONFIG)
cursor = conn.cursor()

upsert_query = """
    INSERT INTO raw.krw_exchange_rates (
        date, stat_code, item_code, item_name, cycle, unit_name, exchange_rate
    ) VALUES %s
    ON CONFLICT (date, stat_code, item_code) 
    DO UPDATE SET 
        item_name = EXCLUDED.item_name,
        cycle = EXCLUDED.cycle,
        unit_name = EXCLUDED.unit_name,
        exchange_rate = EXCLUDED.exchange_rate,
        created_at = CURRENT_TIMESTAMP;
"""

total_inserted = 0

print(f"🚀 [수집 시작] {START_DATE_LIMIT.strftime('%Y-%m-%d')} ~ {NOW.strftime('%Y-%m-%d')} (총 {len(date_ranges)}개 월별 요청)")

# 월별 진행상황 진행바(tqdm)
for start_str, end_str in tqdm(date_ranges, desc="환율 수집 진행률"):
    # API URL (한 번 요청 시 최대 5000건 지정)
    url = f"https://ecos.bok.or.kr/api/{SERVICE_NAME}/{ECOS_KEY}/{REQ_TYPE}/{LANG}/1/5000/{STAT_CODE}/{CYCLE}/{start_str}/{end_str}"

    try:
        res = requests.get(url, timeout=10)
        data = res.json()

        if "StatisticSearch" in data and "row" in data["StatisticSearch"]:
            rows = data["StatisticSearch"]["row"]

            insert_batch = []
            for row in rows:
                # YYYYMMDD -> YYYY-MM-DD 변환
                raw_time = row.get("TIME", "")
                if len(raw_time) == 8:
                    formatted_date = f"{raw_time[:4]}-{raw_time[4:6]}-{raw_time[6:]}"
                else:
                    continue

                # 데이터 정제
                raw_value = row.get("DATA_VALUE", "0")
                try:
                    rate = float(raw_value)
                except ValueError:
                    continue  # 수치 변환 불가 데이터 건너뜀

                insert_batch.append(
                    (
                        formatted_date,
                        row.get("STAT_CODE"),
                        row.get("ITEM_CODE1"),
                        row.get("ITEM_NAME1"),
                        CYCLE,
                        row.get("UNIT_NAME"),
                        rate,
                    )
                )

            # DB Bulk Insert 실행
            if insert_batch:
                execute_values(cursor, upsert_query, insert_batch)
                conn.commit()
                total_inserted += len(insert_batch)

    except Exception as e:
        conn.rollback()
        print(f"\n[오류 발생] {start_str}~{end_str} 수집 중 에러: {e}")

cursor.close()
conn.close()
print(f"\n✅ [수집 완료] 총 {total_inserted}건의 일별 환율 데이터가 raw.krw_exchange_rates 테이블에 저장되었습니다.")
```

```python
# 4. 월별 날짜 구간 생성 함수
def get_monthly_ranges(start_dt, end_dt):
    ranges = []
    current_year = start_dt.year
    current_month = start_dt.month

    while (current_year, current_month) <= (end_dt.year, end_dt.month):
        _, last_day = calendar.monthrange(current_year, current_month)

        s_day = start_dt.day if (current_year == start_dt.year and current_month == start_dt.month) else 1
        e_day = min(last_day, end_dt.day) if (current_year == end_dt.year and current_month == end_dt.month) else last_day

        start_str = f"{current_year}{current_month:02d}{s_day:02d}"
        end_str = f"{current_year}{current_month:02d}{e_day:02d}"

        ranges.append((start_str, end_str))

        if current_month == 12:
            current_year += 1
            current_month = 1
        else:
            current_month += 1

    return ranges
```

```python
date_ranges = get_monthly_ranges(START_DATE_LIMIT, NOW)

conn = psycopg2.connect(**DB_CONFIG)
cursor = conn.cursor()
total_inserted = 0

print(f"🚀 [기준금리 수집 시작] {START_DATE_LIMIT.strftime('%Y-%m-%d')} ~ {NOW.strftime('%Y-%m-%d')} (총 {len(date_ranges)}개 월별 요청)")

# 월별 순회 수집
for start_str, end_str in tqdm(date_ranges, desc="기준금리 수집 진행률"):
    url = f"https://ecos.bok.or.kr/api/{SERVICE_NAME}/{ECOS_KEY}/{REQ_TYPE}/{LANG}/1/5000/{STAT_CODE}/{CYCLE}/{start_str}/{end_str}"

    try:
        res = requests.get(url, timeout=10)
        data = res.json()

        if "StatisticSearch" in data and "row" in data["StatisticSearch"]:
            rows = data["StatisticSearch"]["row"]
            insert_batch = []

            for row in rows:
                raw_time = row.get("TIME", "")  # YYYYMMDD
                if len(raw_time) == 8:
                    formatted_date = f"{raw_time[:4]}-{raw_time[4:6]}-{raw_time[6:]}"
                else:
                    continue

                # 데이터 정제 (None 및 빈 값 체크)
                raw_value = row.get("DATA_VALUE")
                if raw_value is None or str(raw_value).strip() == "":
                    continue

                try:
                    base_rate = float(raw_value)
                except (ValueError, TypeError):
                    continue

                insert_batch.append((
                    formatted_date,
                    row.get("STAT_CODE"),
                    row.get("STAT_NAME"),
                    row.get("ITEM_CODE1"),
                    row.get("ITEM_NAME1"),
                    row.get("UNIT_NAME"),
                    base_rate
                ))

            if insert_batch:
                execute_values(cursor, UPSERT_QUERY, insert_batch)
                conn.commit()
                total_inserted += len(insert_batch)

    except Exception as e:
        conn.rollback()
        print(f"\n[오류] {start_str}~{end_str} 데이터 수집 실패: {e}")

cursor.close()
conn.close()
print(f"\n✅ [수집 완료] 총 {total_inserted}건의 일별 기준금리 데이터가 raw.bok_base_rates에 적재되었습니다.")
```
